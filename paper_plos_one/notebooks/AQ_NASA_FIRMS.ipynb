## Find nearest location

### Subtask:
For each fire incident, find the location with the minimum distance.


**Reasoning**:
For each row in `fire_incidents_daily`, find the index of the minimum distance in the 'distances' column and use that index to find the corresponding location name from `location_mapping`. Store this location name in a new column.



````python
fire_incidents_with_location = fire_incidents_daily.copy()
fire_incidents_with_location['location_name'] = fire_incidents_with_location['distances'].apply(
    lambda x: location_mapping.loc[x.index(min(x)), 'location_name']
)
fire_incidents_with_location['min_distance'] = fire_incidents_with_location['distances'].apply(
    lambda x: min(x)
)
display(fire_incidents_with_location.head())
````
